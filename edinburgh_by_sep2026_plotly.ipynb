```python
import pandas as pd
```

```python
pd.options.plotting.backend = "plotly"
```

```python
craigmillar_portobello=pd.read_html('https://www.edinburgh.gov.uk/local-government-election-4/local-government-election-results-1')
```

```python
craigmillar_portobello[0]
```

```python
(craigmillar_portobello[0]
 .assign(candidate_party=craigmillar_portobello[0].Candidate+craigmillar_portobello[0].Party)
 .set_index(['candidate_party'])
 .sort_values('First preference votes')
 ['First preference votes']
 .plot(kind='barh')
)
```
