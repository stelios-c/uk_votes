# Edinburgh local government by-election results September 2026

```python
import pandas as pd
```

```python
pd.options.plotting.backend = "matplotlib"
```

```python
craigmillar_portobello=pd.read_html('https://www.edinburgh.gov.uk/local-government-election-4/local-government-election-results-1')
```

```python
craigmillar_portobello[0]
```

```python
craigmillar_portobello[0].set_index(['Candidate','Party']).sort_values('First preference votes').plot(kind='barh')
```
